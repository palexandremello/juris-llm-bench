## 3. Rodar o exemplo

Tudo o que define o benchmark está em `exemplos/decisoes/config.yaml`: modelos, campos e prompt.
O gabarito está em `exemplos/decisoes/gabarito.csv`. Na primeira vez, cada modelo é baixado (~2-3 GB cada).
Se a sessão cair, rode a célula de novo: ela continua de onde parou.

```
from juris_llm_bench import carregar_config, rodar, avaliar

cfg = carregar_config("exemplos/decisoes/config.yaml")
rodar(cfg)
r = avaliar(cfg)
r["resumo"]
```

## 4. Resultados
Acerto de cada modelo em cada campo:

```
r["por_campo"].style.format("{:.0%}").background_gradient(cmap="RdYlGn", vmin=0, vmax=1)
```

Onde cada modelo errou (previsto × esperado):

```
r["erros"]
```

## 5. Com os seus documentos

1. Copie `exemplos/decisoes` para uma pasta nova (ex.: no Google Drive).
2. Em `config.yaml`, aponte `textos:` para a pasta de saída do juris-pdf e ajuste `campos` e `prompt`.
3. Rode sem gabarito: as respostas de cada modelo ficam em `resultados/previsoes_<modelo>.csv`.
4. Abra uma delas no Excel ou no Google Sheets, corrija, e salve como `gabarito.csv`.
5. Rode `avaliar(cfg)` de novo para recalcular as notas sem chamar os modelos.
